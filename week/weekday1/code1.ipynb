## 张量运算(广播)

```python
a = np.arange(6).reshape(2, 3)+1
b = np.arange(3)+1
c = np.arange(2).reshape(2, 1)+1

a, b, a*b, a*c, a+b # 广播机制
```

```python
print(b,c,b+c,sep='\n----------------\n')
```
